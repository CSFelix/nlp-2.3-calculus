<h1 id='0-friezas-power-level-growth' style='color:#7159c1; border-bottom:3px solid #7159c1; letter-spacing:2px; font-family:JetBrains Mono; font-weight: bold; text-align:left; font-size:240%;padding:0'>📖 | Frieza's Power Level Growth</h1>

Frieza is training in his Hyperbolic Time Chamber to surpass Goku and Vegeta. His power level at any time $\mathbf{t}$ (in months) is given by the function:

$$
\mathbf{P}(\mathbf{t}) = 5000\mathbf{e}^{0.20\mathbf{t}}
$$

where:

- $\mathbf{P}(\mathbf{t})$ is Frieza's power level at any time $\mathbf{t}$;
- $\mathbf{e}$ is Euler's number ($\approx 2.718$), meaning his power grows `exponentially` over time;
- $0.20$ is the `growth rate`, meaning that Frieza gets strong `really fast`.

Now, let's find:

1. How fast Frieza's power level is increasing (Derivative of $\mathbf{P}(\mathbf{t})$);
2. Total power gained over a period (Integral of $\mathbf{P}(\mathbf{t})$).

---

**- Step 1: Find the Rate of Power Growth (Derivative of $\mathbf{P}(\mathbf{t}$)**

Since the derivative tells us the rate of change, we differentiate $\mathbf{P}(\mathbf{t}):

$$
\mathbf{P'}(\mathbf{t}) = \frac{\mathbf{d}}{\mathbf{dt}} (5000\mathbf{e}^{0.20\mathbf{t}})
$$

Using the `Exponential Derivative Rule` $\frac{\mathbf{d}}{\mathbf{dt}} (\mathbf{e}^{\alpha \mathbf{x}}) = \alpha \mathbf{e}^{\alpha \mathbf{x}}$:

$$
\begin{array}
    \mathbf{P'}(\mathbf{t}) = 5000 \cdot 0.20\mathbf{e}^{0.20\mathbf{t}} \\ \\
    \mathbf{P'}(\mathbf{t}) = 1000 \mathbf{e}^{0.20\mathbf{t}}
\end{array}
$$

Interpretation:

- at $\mathbf{t} = 0$, the beginning of the training, Frieza is gaining `1000 power per month` at first:

$$
\mathbf{P'}(0) = 1000 \mathbf{e}^{0.20 \cdot 0} = 1000 \mathbf{e}^0 = 1000 \cdot 1 = 1000
$$

- at $\mathbf{t} = 6$, six months of training, Frieza is gaining `3320 power per month` at first:

$$
\mathbf{P'}(6) = 1000 \mathbf{e}^{0.20 \cdot 6} = 1000 \mathbf{e}^1.20 \approx 1000 \cdot 3.32 \approx 3320
$$

This means his power is `snowballing` — each month he gains `even more` than before!
