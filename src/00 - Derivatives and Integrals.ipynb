<h1 id='0-derivatives' style='color:#7159c1; border-bottom:3px solid #7159c1; letter-spacing:2px; font-family:JetBrains Mono; font-weight: bold; text-align:left; font-size:240%;padding:0'>📊 | Derivatives</h1>

`Derivatives` tell how much something is changing, that is, they tell the `slope` of a function. For instance, picture that Krillin, from Dragon Ball Z, is flying from Kame House to Corp. Capsula and we desire to know how fast is he going at any moment during the flying. So, we can conclude that Krillin's speed is the derivative of his position - it tells us how his position changes over time.

<h1 id='1-integrals' style='color:#7159c1; border-bottom:3px solid #7159c1; letter-spacing:2px; font-family:JetBrains Mono; font-weight: bold; text-align:left; font-size:240%;padding:0'>🏝️ | Integrals</h1>

In the other hand, `Integrals` are the "inverse" of Derivatives, telling how much something changed over the whole function instead of a specific change at a specific moment. Using the same example of Krillin, consider the Integral as the whole distance the character flew from Kame House to Corp Capsula - it tells us the total distance travelled `adding up` all the Derivatives over time.

<h1 id='2-derivatives-rules' style='color:#7159c1; border-bottom:3px solid #7159c1; letter-spacing:2px; font-family:JetBrains Mono; font-weight: bold; text-align:left; font-size:240%;padding:0'>📊 | Derivatives Rules</h1>

We will go further into more examples and solutions in the next notebooks, for now, let's list some rules that will help us to solve some calculations better.

> 1. Power Rule:

$$
\frac{\mathbf{d}}{\mathbf{dx}} \mathbf{x}^n = \mathbf{nx}^{\mathbf{n} - 1}
$$

> 2. Sum Rule:

$$
\frac{\mathbf{d}}{\mathbf{dx}} [\mathbf{f}(\mathbf{x}) + \mathbf{g}(\mathbf{x})] = \mathbf{f'}(\mathbf{x}) + \mathbf{g'}(\mathbf{x})
$$

> 3. Constant Rule:

$$
\frac{\mathbf{d}}{\mathbf{dx}} (\mathbf{C}) = 0
$$

> 4. Constant Multiple Rule:

$$
\frac{\mathbf{d}}{\mathbf{dx}} [\mathbf{C} \mathbf{f}(\mathbf{x})] = \mathbf{C} \mathbf{f'}(\mathbf{x}) 
$$

> 5. Product Rule:

$$
\frac{\mathbf{d}}{\mathbf{dx}} [\mathbf{f}(\mathbf{x}) \cdot \mathbf{g}(\mathbf{x})] = \mathbf{f'}(\mathbf{x}) \cdot \mathbf{g}(\mathbf{x}) + \mathbf{f}(\mathbf{x}) \cdot \mathbf{g'}(\mathbf{x})
$$

> 6. Quotient Rule:

$$
\frac{\mathbf{d}}{\mathbf{dx}} \left[\frac{\mathbf{f}(\mathbf{x})}{\mathbf{g}(\mathbf{x})}\right] =
\frac
{\mathbf{f'}(\mathbf{x}) \cdot \mathbf{g}(\mathbf{x}) - \mathbf{f}(\mathbf{x}) \cdot \mathbf{g'}(\mathbf{x})}
{\mathbf{g}(\mathbf{x}^2)}
$$

> 7. Chain Rule:

$$
\frac{\mathbf{d}}{\mathbf{dx}} \mathbf{f}(\mathbf{g}(\mathbf{x})) = \mathbf{f'}(\mathbf{g}(\mathbf{x})) \cdot \mathbf{g'}(\mathbf{x})
$$

<h1 id='3-integrals-rules' style='color:#7159c1; border-bottom:3px solid #7159c1; letter-spacing:2px; font-family:JetBrains Mono; font-weight: bold; text-align:left; font-size:240%;padding:0'>🏝️ | Integrals Rules</h1>

> 1. Indefinite Integral Rule - used to find a General Function from Derivative:

$$
\int \mathbf{f}(\mathbf{x}) \mathbf{dx}
$$

> 2. Definite Integral Rule - used to find the area under the curve from $\mathbf{x} = \mathbf{a}$ to $\mathbf{y} = \mathbf{b}$:

$$
\int_a^b \mathbf{f}(\mathbf{x}) \mathbf{dx}
$$
