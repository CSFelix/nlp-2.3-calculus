<h1 id='0-gokus-speed-and-distance' style='color:#7159c1; border-bottom:3px solid #7159c1; letter-spacing:2px; font-family:JetBrains Mono; font-weight: bold; text-align:left; font-size:240%;padding:0'>📖 | Goku's Speed & Distance</h1>

Goku, from Dragon Ball series, is flying to train on King Kai’s planet, but he's got a problem — he's not sure how far he will travel or how fast he’s accelerating. His speed at any time $\mathbf{t}$ (in seconds) is given by the function:

$$
\mathbf{v}(\mathbf{t}) = 5\mathbf{t}^2 + 2\mathbf{t} + 3
$$

where $\mathbf{v}(\mathbf{t})$ is his velocity in meters per second (m/s).

Now, we want to find:

1. Goku’s acceleration at any time $\mathbf{t}$ (derivative of velocity;
2. Goku’s total distance traveled (integral of velocity).

---

**- Step 1) Find Acceleration (Derivative of Velocity)**

Since acceleration is the rate of change of velocity, we find the derivative of $\mathbf{v}(\mathbf{t})$:

$$
\alpha(\mathbf{t}) = \frac{\mathbf{d}}{\mathbf{dt}} (5\mathbf{t}^2 + 2\mathbf{t} + 3)
$$

Using the `Power Rule` $\frac{\mathbf{d}}{\mathbf{dt}} \left( \mathbf{t}^\mathbf{n} \right) = \mathbf{nt}^{\mathbf{n} - 1}$:

$$
\frac{\mathbf{d}}{\mathbf{dt}} (5\mathbf{t}^2) = 5 \cdot 2\mathbf{t}^{2-1} = 10\mathbf{t}
$$

$$
\frac{\mathbf{d}}{\mathbf{dt}} (2\mathbf{t}) = 2 \cdot 1\mathbf{t}^{1-1} = 2
$$

$$
\frac{\mathbf{d}}{\mathbf{dt}} (3) = 0
$$

So, Goku's acceleration is:

$$
\alpha(\mathbf{t}) = 10\mathbf{t} + 2
$$

Interpretation:

- at $\mathbf{t} = 0$, Goku's acceleration is $2 \mathbf{m} / \mathbf{s}^2$;
- at $\mathbf{t} = 5$, Goku's acceleration is $10 \cdot 5 + 2 = 52 \mathbf{m} / \mathbf{s}^2$.

This means the `longer` Goku flies, the `faster` he speeds up!

---

**- Step 2) Find Total Distance Traveled (Integral of Velocity)**

Now, we want to know how far Goku travels by finding the integral of $\mathbf{v}(\mathbf{t})$:

$$
\mathbf{s}(\mathbf{t}) = \int \mathbf{v}(\mathbf{t}) \mathbf{dt} = \int (5\mathbf{t}^2 + 2\mathbf{t} + 3) \mathbf{dt}
$$

Using the `Power Rule` for integration $\int \mathbf{t}^\mathbf{n} \mathbf{dt} = \frac{\mathbf{t}^(\mathbf{n + 1})}{\mathbf{n} + 1}$:

$$
\int 5\mathbf{t}^2 \mathbf{dt} = 5 \cdot \frac{\mathbf{t}^{2+1}}{2 + 1} = 5 \cdot \frac{\mathbf{t}^3}{3} = \frac{5}{3} \mathbf{t}^3
$$

$$
\int 2\mathbf{t} \mathbf{dt} = 2 \cdot \frac{\mathbf{t}^{1+1}}{1 + 1} = 2 \cdot \frac{\mathbf{t}^2}{2} = \mathbf{t}^2
$$

$$
\int 3 \mathbf{dt} = 3\mathbf{t}
$$

$$
\int \mathbf{dt} = \mathbf{C}
$$

So, Goku's `position function` (total distance traveled) is:

$$
\mathbf{s}(\mathbf{t}) = \frac{5}{3}\mathbf{t}^3 + \mathbf{t}^2 + 3\mathbf{t} + \mathbf{C}
$$

---

**- Step 3) Solve for $\mathbf{C}$ (Initial Condition)**

Let's assume that Goku starts at position 0 when $\mathbf{t} = 0$, meaning $\mathbf{s}(0) = 0$, then:

$$
0 = \frac{5}{3}(0)^3 + (0)^2 + 3 \cdot (0) + \mathbf{C}
$$

$$
\mathbf{C} = 0
$$

Since $\mathbf{C}$ is zero, we can remove it from the `position function (integral)` equation!

---

**- Step 4) Final Answer**

> Acceleration:

$$
\alpha(\mathbf{t}) = 10\mathbf{t} + 2
$$

> Total Distance Traveled:

$$
\mathbf{s}(\mathbf{t}) = \frac{5}{3}\mathbf{t}^3 + \mathbf{t}^2 + 3\mathbf{t}
$$

> Interpretation:

If Goku fliest for $\mathbf{t} = 4 \text{ seconds}$, he travels:

$$
\begin{array}
    \mathbf{s}(4) = \frac{5}{3} \cdot (4)^3 + (4)^2 + 3 \cdot (4) = \\ \\
    \frac{5}{3} \cdot (64) + 16 + 12 = \\ \\
    106.67 + 16 + 12 \approx \\ \\
    134.67
\end{array}
$$

After 4 seconds, Goku has flown about `134.67 meters`!
